```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


NHA=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Accessories.csv')
NHB=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Bags.csv')
NHBo=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Bottoms.csv')
NHCD=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Ceiling Decor.csv')
NHCO=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Clothing Other.csv')
NHDU=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Dress-Up.csv')
NHF=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Fencing.csv')
NHFl=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Floors.csv')
NHHw=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Headwear.csv')
NHHwa=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Housewares.csv')
NHIS=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Interior Structures.csv')
NHM=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Miscellaneous.csv')
NHP=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Photos.csv')
NHR=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Rugs.csv')
NHS=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Shoes.csv')
NHSo=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Socks.csv')
NHT=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Tools_Goods.csv')
NHTo=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Tops.csv')
NHU=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Umbrellas.csv')
NHWm=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Wall-mounted.csv')
NHW=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Wallpaper.csv')

NHFrames=[NHA,NHB,NHBo,NHCD,NHCO,NHDU,NHF,NHFl,NHHw,NHHwa,NHIS,NHM,NHP,NHR,NHS,NHSo,NHT,NHTo,NHU,NHWm,NHW]
NewHorizonsData=pd.concat(NHFrames)
NewHorizonsDIY=NewHorizonsData[['Name','DIY']]

NLA=pd.read_csv('Data Spreadsheet for Animal Crossing_ New Leaf - Accessories.csv')
NLB=pd.read_csv('Data Spreadsheet for Animal Crossing_ New Leaf - Bottoms.csv')
NLD=pd.read_csv('Data Spreadsheet for Animal Crossing_ New Leaf - Dresses.csv')
NLE=pd.read_csv('Data Spreadsheet for Animal Crossing_ New Leaf - Equipment.csv')
NLF=pd.read_csv('Data Spreadsheet for Animal Crossing_ New Leaf - Flooring.csv')
NLFu=pd.read_csv('Data Spreadsheet for Animal Crossing_ New Leaf - Furniture.csv')
NLH=pd.read_csv('Data Spreadsheet for Animal Crossing New Horizons - Fencing.csv')
NLHHD=pd.read_csv('Data Spreadsheet for Animal Crossing_ New Leaf - HHD.csv')
NLO=pd.read_csv('Data Spreadsheet for Animal Crossing_ New Leaf - Other Items.csv')
NLP=pd.read_csv('Data Spreadsheet for Animal Crossing_ New Leaf - Pictures.csv')
NLS=pd.read_csv('Data Spreadsheet for Animal Crossing_ New Leaf - Socks_Shoes.csv')
NLSt=pd.read_csv('Data Spreadsheet for Animal Crossing_ New Leaf - Stationery.csv')
NLT=pd.read_csv('Data Spreadsheet for Animal Crossing_ New Leaf - Tops.csv')
NLU=pd.read_csv('Data Spreadsheet for Animal Crossing_ New Leaf - Umbrellas.csv')
NLW=pd.read_csv('Data Spreadsheet for Animal Crossing_ New Leaf - Wallpaper.csv')
NLWS=pd.read_csv('Data Spreadsheet for Animal Crossing_ New Leaf - Wet Suits.csv')

NLFrames=[NLA,NLB,NLD,NLE,NLF,NLFu,NLH,NLHHD,NLO,NLP,NLS,NLSt,NLT,NLU,NLW,NLWS]
NewLeafData=pd.concat(NLFrames)
NewLeafNames=NewLeafData[['Name']]

NHFilterdNot=~NewHorizonsDIY['Name'].isin(NewLeafNames['Name'])
NHFilterdNot=NewHorizonsDIY[NHFilterdNot]
NHFilterdYes=NewHorizonsDIY['Name'].isin(NewLeafNames['Name'])
NHFilterdYes=NewHorizonsDIY[NHFilterdYes]
#NHFilterd = NewHorizonsDIY[NewHorizonsDIY['Name'] ==NewLeafNames['Name']] 

mask=NHFilterdYes['DIY']=='Yes'
NHYesDIYYes=NHFilterdYes[mask]
NHYesDIYYesSum=len(NHYesDIYYes)
mask=NHFilterdYes['DIY']=='No'
NHYesDIYNo=NHFilterdYes[mask]
NHYesDIYNoSum=len(NHYesDIYNo)
mask=NHFilterdNot['DIY']=='Yes'
NHNoDIYYes=NHFilterdNot[mask]
NHNoDIYYesSum=len(NHNoDIYYes)
mask=NHFilterdNot['DIY']=='No'
NHNoDIYNo=NHFilterdNot[mask]
NHNoDIYNoSum=len(NHNoDIYNo)
Title=['Craftable','Not Craftable']
dataNL=[NHYesDIYYesSum,NHYesDIYNoSum]
datanoNL=[NHNoDIYYesSum,NHNoDIYNoSum]
T1='Old Items'
T2='New Items'
fig, (AI,OI, NI) = plt.subplots(1, 3, figsize=(10, 5))


mask=NewHorizonsDIY['DIY']=='Yes'
NHAY=NewHorizonsDIY[mask]
NHAYSum=len(NHAY)
mask=NewHorizonsDIY['DIY']=='No'
NHAN=NewHorizonsDIY[mask]
NHANSum=len(NHAN)
dataAll=[NHAYSum,NHANSum]
T3='All New Horizons Items'

AI.pie(dataAll, labels=Title, autopct='%1.1f%%', startangle=90)
AI.set_title(T3)

OI.pie(dataNL, labels=Title, autopct='%1.1f%%', startangle=90)
OI.set_title(T1)

NI.pie(datanoNL, labels=Title, autopct='%1.1f%%', startangle=90)
NI.set_title(T2)

plt.show()
```

```python
NewHorizonsCost=NewHorizonsData[['Buy']]
NewLeafCost=NewLeafData[['Buy Price (Bells)']]
NewLeafCost=NewLeafCost.dropna()
mask=NewLeafCost['Buy Price (Bells)']!='-'
NewLeafCost=NewLeafCost[mask]
mask=NewLeafCost['Buy Price (Bells)']!='Varies'
NewLeafCost=NewLeafCost[mask]
NewHorizonsCost=NewHorizonsCost.dropna()

NHMFilterd=NewHorizonsCost['Buy'].isin(NewLeafCost['Buy Price (Bells)'])
NHMFilterd=NewHorizonsCost[NHMFilterd]

NLMFilterd=NewLeafCost['Buy Price (Bells)'].isin(NewHorizonsCost['Buy'])
NLMFilterd=NewLeafCost[NLMFilterd]
NewLeafCost['Buy Price (Bells)']=NewLeafCost['Buy Price (Bells)'].str.replace(',', '')
NewLeafCost['Buy Price (Bells)']=NewLeafCost['Buy Price (Bells)'].astype(int)

NLCSum=NewLeafCost['Buy Price (Bells)'].sum()

mask=NewHorizonsCost['Buy']!='NFS'
NewHorizonsCost=NewHorizonsCost[mask]
NewHorizonsCost['Buy']=NewHorizonsCost['Buy'].astype(int)
NHCSum=NewHorizonsCost['Buy'].sum()


mask=NHMFilterd['Buy']!='NFS'
NHMFilterd=NHMFilterd[mask]
NHMFilterd['Buy']=NHMFilterd['Buy'].astype(int)
NHCFSum=NHMFilterd['Buy'].sum()

NLMFilterd['Buy Price (Bells)']=NLMFilterd['Buy Price (Bells)'].str.replace(',', '')
NLMFilterd['Buy Price (Bells)']=NLMFilterd['Buy Price (Bells)'].astype(int)
NLMFSum=NLMFilterd['Buy Price (Bells)'].sum()

values=[NLCSum/len(NewLeafCost),NLMFSum/len(NLMFilterd),NHCSum/len(NewHorizonsCost),NHCFSum/len(NHMFilterd)]

fig, ax = plt.subplots()
y_pos = np.arange(len(values))
ax.barh(y_pos, values, align='center')
measure=['New Leaf Items','New Leaf Items that are in New Horizon', 'New Horizon Items','New Horizon Items that were in New Leaf']
ax.set_yticks(y_pos, labels=measure)
ax.set_xlabel('Average Bell Cost of Items')
ax.set_title('How Much is the Avearge Cost of Items in New Horizons and New Leaf')
plt.show()
```
